```python
import pandas as pd

df1 = pd.read_csv("youtube_comments_AI_2020_2023.csv")
df2 = pd.read_csv("youtube_comments_AI.csv")

df1["period"] = "2020-2023"
df2["period"] = "2022-2026"

df = pd.concat([df1, df2], ignore_index=True)
df.head()
```

```python
# Añadir año del comentario (faltaba en ambos)
df["date"] = pd.to_datetime(df["date"], utc=True)
df["comment_year"] = df["date"].dt.year

# Clasificar por tema
def clasificar_tema(query):
    educacion = ["education", "teacher", "school", "homework", "learning", "student"]
    empleo = ["job", "work", "employ", "unemploy", "worker", "replace worker"]
    if any(p in query.lower() for p in educacion):
        return "education"
    elif any(p in query.lower() for p in empleo):
        return "employment"
    else:
        return "general"

df["topic"] = df["query_used"].apply(clasificar_tema)

# Eliminar duplicados que puedan existir entre los dos datasets
df = df.drop_duplicates(subset="comment_id")

df.to_csv("youtube_comments_AI_FULL.csv", index=False)
print(f"Total comentarios: {len(df)}")
print(df["topic"].value_counts())
print(df["comment_year"].value_counts().sort_index())
```

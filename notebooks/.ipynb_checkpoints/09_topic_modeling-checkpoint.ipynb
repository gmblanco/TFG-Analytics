## BERTopic

```python
# ============================================================
# 06 — Topic Modeling con BERTopic
# ============================================================
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# --- Carga del dataset con columnas calculadas en 05b ---
df = pd.read_parquet("../data/processed/youtube_sentiment_v2.parquet")
df["date"] = pd.to_datetime(df["date"], utc=True, errors="coerce")

# --- Recuperar video_title desde el CSV FULL ---
df_full = pd.read_csv("youtube_comments_AI_FULL.csv",
                      usecols=["video_id", "video_title"])

video_titles = df_full.drop_duplicates("video_id")

df = df.merge(video_titles, on="video_id", how="left")

# Verificar
n_nulos = df["video_title"].isna().sum()
print(f"Comentarios cargados:       {len(df):,}")
print(f"Sin video_title tras merge: {n_nulos:,} ({n_nulos/len(df)*100:.1f}%)")

# --- Verificar columnas clave ---
assert "fase"              in df.columns, "Falta columna 'fase'"
assert "sector_education"  in df.columns, "Falta columna 'sector_education'"
assert "sector_employment" in df.columns, "Falta columna 'sector_employment'"
assert "video_title"       in df.columns, "Falta columna 'video_title'"

print("\nColumnas clave verificadas correctamente")
```

```python
# Instalación — ejecutar una sola vez
import subprocess
subprocess.run(["pip", "install", "bertopic", "sentence-transformers", 
                "umap-learn", "hdbscan", "--quiet"])
```

```python
# ============================================================
# PREPARACIÓN DE CORPUS — con video_title disponible
# ============================================================

MIN_PALABRAS = 10

def preparar_corpus(df, sector_col, sentimiento):
    subset = df[
        (df[sector_col] == True) &
        (df["sentiment_label"] == sentimiento) &
        (df["n_palabras_clean"] >= MIN_PALABRAS)
    ].copy().reset_index(drop=True)
    
    textos = subset["text_clean"].tolist()
    return textos, subset

textos_emp_neg, df_emp_neg = preparar_corpus(df, "sector_employment", "negative")
textos_emp_pos, df_emp_pos = preparar_corpus(df, "sector_employment", "positive")
textos_edu_neg, df_edu_neg = preparar_corpus(df, "sector_education",  "negative")
textos_edu_pos, df_edu_pos = preparar_corpus(df, "sector_education",  "positive")

# Verificar que video_title está disponible
print(df_emp_neg["video_title"].value_counts().head(3))
```

```python
# ============================================================
# CONFIGURACIÓN Y EJECUCIÓN DE BERTOPIC
# ============================================================

import spacy
from bertopic import BERTopic
from bertopic.vectorizers import ClassTfidfTransformer
from sentence_transformers import SentenceTransformer
from umap import UMAP
from hdbscan import HDBSCAN
from sklearn.feature_extraction.text import CountVectorizer

# ------------------------------------
# 1. MODELOS BASE
# ------------------------------------

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
nlp = spacy.load("en_core_web_sm")

# ------------------------------------
# 2. STOPWORDS EXTENDIDAS
# ------------------------------------

EXTRA_STOPWORDS = {
    # Opinión genérica
    "like", "just", "know", "think", "thing", "going", "want",
    "need", "make", "way", "say", "really", "good", "great",
    "right", "mean", "help", "better", "well", "true", "sure",
    "feel", "lot", "bit", "actually", "probably", "basically",
    # Interacción con YouTube
    "video", "watch", "youtube", "comment", "channel",
    "thanks", "thank", "subscribe", "share", "view",
    # Tiempo genérico
    "time", "day", "year", "years", "ago", "today", "now",
    # Referencias genéricas
    "people", "person", "someone", "everyone", "anybody",
    # IA — está en todos los tópicos, no discrimina
    "ai", "artificial", "intelligence", "machine",
    # Verbos auxiliares frecuentes
    "use", "using", "used", "get", "got", "go", "come",
    "see", "look", "said", "take", "give", "put", "run",
    "work", "try", "start", "end", "move", "turn", "keep"
}

# ------------------------------------
# 3. FUNCIÓN DE LEMATIZACIÓN
# ------------------------------------

def lematizar(texto):
    doc = nlp(str(texto).lower()[:1000])
    lemas = [
        token.lemma_
        for token in doc
        if not token.is_stop
        and not token.is_punct
        and not token.like_num
        and not token.is_space
        and token.is_alpha
        and len(token.lemma_) > 2
        and token.lemma_ not in EXTRA_STOPWORDS
    ]
    return " ".join(lemas)

# ------------------------------------
# 4. LEMATIZACIÓN
# ------------------------------------

print("Lematizando corpus... (puede tardar 5-8 minutos)")
textos_emp_neg_lem = [lematizar(t) for t in textos_emp_neg]
textos_emp_pos_lem = [lematizar(t) for t in textos_emp_pos]
textos_edu_neg_lem = [lematizar(t) for t in textos_edu_neg]
textos_edu_pos_lem = [lematizar(t) for t in textos_edu_pos]
print("Lematización completada\n")

for nombre, textos_lem in [
    ("emp_neg", textos_emp_neg_lem),
    ("emp_pos", textos_emp_pos_lem),
    ("edu_neg", textos_edu_neg_lem),
    ("edu_pos", textos_edu_pos_lem),
]:
    longitudes = [len(t.split()) for t in textos_lem]
    print(f"{nombre}: media={sum(longitudes)/len(longitudes):.1f} palabras | "
          f"vacíos={sum(1 for l in longitudes if l == 0)}")

# ------------------------------------
# 5. PARÁMETROS POR CORPUS
# ------------------------------------

# emp_neg: corpus grande y heterogéneo — parámetros estándar
# emp_pos, edu_pos: corpus pequeños (~2k) — n_neighbors bajo
#   para que UMAP preserve estructura local, min_cluster_size
#   bajo para encontrar grupos pequeños
# edu_neg: corpus medio pero homogéneo (nube continua confirmada
#   visualmente) — n_neighbors muy bajo y cluster_selection_method
#   "leaf" para forzar separación de grupos difusos

CONFIGS = {
    "emp_neg": {
        "umap":    dict(n_components=5, n_neighbors=15, min_dist=0.0,
                        metric="cosine", random_state=123),
        "hdbscan": dict(min_cluster_size=50, min_samples=10,
                        metric="euclidean", cluster_selection_method="eom",
                        prediction_data=True),
        "min_df":  5,
    },
    "emp_pos": {
        "umap":    dict(n_components=5, n_neighbors=5, min_dist=0.0,
                        metric="cosine", random_state=123),
        "hdbscan": dict(min_cluster_size=15, min_samples=3,
                        metric="euclidean", cluster_selection_method="leaf",
                        prediction_data=True),
        "min_df":  2,
    },
    "edu_neg": {
        "umap":    dict(n_components=5, n_neighbors=5, min_dist=0.0,
                        metric="cosine", random_state=123),
        "hdbscan": dict(min_cluster_size=15, min_samples=3,
                        metric="euclidean", cluster_selection_method="leaf",
                        prediction_data=True),
        "min_df":  2,
    },
    "edu_pos": {
        "umap":    dict(n_components=5, n_neighbors=5, min_dist=0.0,
                        metric="cosine", random_state=123),
        "hdbscan": dict(min_cluster_size=15, min_samples=3,
                        metric="euclidean", cluster_selection_method="leaf",
                        prediction_data=True),
        "min_df":  2,
    },
}

# ------------------------------------
# 6. EJECUCIÓN DE BERTOPIC
# ------------------------------------

# Todos los componentes se instancian DENTRO de la función
# para garantizar estado limpio en cada ejecución.
# Solo embedding_model se comparte — no tiene estado mutable.

def ejecutar_bertopic(textos_lem, nombre, config_key):
    print(f"\nProcesando: {nombre} ({len(textos_lem):,} comentarios)...")
    cfg = CONFIGS[config_key]

    vectorizer = CountVectorizer(
        min_df=cfg["min_df"],
        ngram_range=(1, 2),
        token_pattern=r"(?u)\b[a-zA-Z]{3,}\b"
    )
    ctfidf = ClassTfidfTransformer(bm25_weighting=True)

    topic_model = BERTopic(
        embedding_model=embedding_model,
        umap_model=UMAP(**cfg["umap"]),
        hdbscan_model=HDBSCAN(**cfg["hdbscan"]),
        vectorizer_model=vectorizer,
        ctfidf_model=ctfidf,
        verbose=False
    )

    topics, probs = topic_model.fit_transform(textos_lem)

    info = topic_model.get_topic_info()
    n_topicos = len(info[info["Topic"] != -1])
    n_ruido   = info[info["Topic"] == -1]["Count"].values[0] \
                if -1 in info["Topic"].values else 0

    print(f"  Tópicos encontrados: {n_topicos}")
    print(f"  Ruido (sin tópico):  {n_ruido:,} ({n_ruido/len(textos_lem)*100:.1f}%)")

    return topic_model, topics, probs

# ------------------------------------
# 7. EJECUCIÓN
# ------------------------------------

# AVISO: puede tardar 20-30 minutos en total
model_emp_neg, topics_emp_neg, _ = ejecutar_bertopic(
    textos_emp_neg_lem, "Empleo — Negativos",    "emp_neg")
model_emp_pos, topics_emp_pos, _ = ejecutar_bertopic(
    textos_emp_pos_lem, "Empleo — Positivos",    "emp_pos")
model_edu_neg, topics_edu_neg, _ = ejecutar_bertopic(
    textos_edu_neg_lem, "Educación — Negativos", "edu_neg")
model_edu_pos, topics_edu_pos, _ = ejecutar_bertopic(
    textos_edu_pos_lem, "Educación — Positivos", "edu_pos")

print("\nTodos los modelos ejecutados correctamente")
```

```python
# Ajuste fino — buscamos 8-20 tópicos por corpus
CONFIGS["emp_pos"]["hdbscan"]["min_cluster_size"] = 40
CONFIGS["edu_neg"]["hdbscan"]["min_cluster_size"] = 70
CONFIGS["edu_pos"]["hdbscan"]["min_cluster_size"] = 50

# Reejecutar solo los tres corpus problemáticos
model_emp_pos, topics_emp_pos, _ = ejecutar_bertopic(
    textos_emp_pos_lem, "Empleo — Positivos",    "emp_pos")
model_edu_neg, topics_edu_neg, _ = ejecutar_bertopic(
    textos_edu_neg_lem, "Educación — Negativos", "edu_neg")
model_edu_pos, topics_edu_pos, _ = ejecutar_bertopic(
    textos_edu_pos_lem, "Educación — Positivos", "edu_pos")
```

```python
# ============================================================
# DIAGNÓSTICO — qué contienen realmente los cuatro corpus
# ============================================================

for nombre, textos_lem in [
    ("emp_neg", textos_emp_neg_lem),
    ("emp_pos", textos_emp_pos_lem),
    ("edu_neg", textos_edu_neg_lem),
    ("edu_pos", textos_edu_pos_lem),
]:
    longitudes = [len(t.split()) for t in textos_lem]
    n_vacios = sum(1 for l in longitudes if l == 0)
    print(f"\n{nombre}:")
    print(f"  N documentos:   {len(textos_lem):,}")
    print(f"  Media palabras: {sum(longitudes)/len(longitudes):.1f}")
    print(f"  Vacíos:         {n_vacios}")
    print(f"  Ejemplo [0]:    {textos_lem[0][:120]}")
    print(f"  Ejemplo [1]:    {textos_lem[1][:120]}")
```

```python
# ¿Cuánto se solapan los corpus?
set_emp_neg = set(textos_emp_neg)
set_emp_pos = set(textos_emp_pos)
set_edu_neg = set(textos_edu_neg)
set_edu_pos = set(textos_edu_pos)

print(f"emp_neg ∩ edu_neg: {len(set_emp_neg & set_edu_neg):,} textos compartidos")
print(f"emp_neg ∩ emp_pos: {len(set_emp_neg & set_emp_pos):,} textos compartidos")
print(f"emp_neg ∩ edu_pos: {len(set_emp_neg & set_edu_pos):,} textos compartidos")
print(f"edu_neg ∩ edu_pos: {len(set_edu_neg & set_edu_pos):,} textos compartidos")
```

```python
import numpy as np
from umap import UMAP

# Generamos embeddings y aplicamos UMAP manualmente
# para ver si el espacio reducido tiene estructura

print("Generando embeddings para edu_neg...")
embeddings_edu_neg = embedding_model.encode(
    textos_edu_neg_lem, 
    show_progress_bar=True,
    batch_size=64
)

print("\nAplicando UMAP...")
umap_test = UMAP(
    n_components=2,  # 2D para poder visualizar
    n_neighbors=10,
    min_dist=0.0,
    metric="cosine",
    random_state=123
)
embeddings_2d = umap_test.fit_transform(embeddings_edu_neg)

print(f"\nForma embeddings originales: {embeddings_edu_neg.shape}")
print(f"Forma embeddings 2D: {embeddings_2d.shape}")

# Ver dispersión — si todos los puntos están muy juntos
# HDBSCAN no puede encontrar clusters
print(f"\nRango X: {embeddings_2d[:,0].min():.3f} → {embeddings_2d[:,0].max():.3f}")
print(f"Rango Y: {embeddings_2d[:,1].min():.3f} → {embeddings_2d[:,1].max():.3f}")

# Visualizar
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 7))
plt.scatter(embeddings_2d[:,0], embeddings_2d[:,1], 
            alpha=0.3, s=5, c="steelblue")
plt.title("UMAP 2D — Educación Negativos\n¿Hay clusters visibles?")
plt.tight_layout()
plt.show()
```

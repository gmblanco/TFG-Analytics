```python
import pandas as pd

df_early = pd.read_csv("youtube_comments_AI_2020_2023.csv")
df_post = pd.read_csv("youtube_comments_AI.csv")
df_2024 = pd.read_csv("youtube_comments_2024.csv")

df_early["period"] = "2020-2023"
df_post["period"] = "2022-2026"
df_2024["period"] = "2024"

df = pd.concat([df_early, df_post, df_2024], ignore_index=True)
df = df.drop_duplicates(subset="comment_id")

df.to_csv("youtube_comments_AI_FULL.csv", index=False)
print(f"Total comentarios: {len(df):,}")

# Verificar distribución temporal
df["date"] = pd.to_datetime(df["date"], utc=True, errors="coerce")
df["comment_year"] = df["date"].dt.year
print(df["comment_year"].value_counts().sort_index())
```
